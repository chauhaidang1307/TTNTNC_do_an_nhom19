Yêu cầu trước khi chạy:
- Đã cài Python **3.10** (kiểm tra ở ô 1).
- Mở notebook với thư mục làm việc là **thư mục gốc của dự án** (nơi có `requirements.txt`).
- Máy có GPU NVIDIA và driver đủ mới nếu muốn dùng CUDA. Không có GPU thì xem ghi chú ở ô 4.

Notebook chỉ gọi Python của `.venv` để cài và kiểm tra, nên không cần kích hoạt venv và không ảnh hưởng Python toàn cục.

## Ô 1. Kiểm tra thư mục, Python 3.10 và GPU

```python
import os
print('Thư mục hiện tại:', os.getcwd())
print('Có requirements.txt:', os.path.exists('requirements.txt'))
!py -3.10 --version
!nvidia-smi
```

Cần thấy `Python 3.10.x` và `Có requirements.txt: True`. Nếu `py -3.10` báo lỗi, hãy cài Python 3.10 trước. Dòng `CUDA Version` ở đầu kết quả `nvidia-smi` là mức CUDA tối đa mà driver hỗ trợ.

## Ô 2. Tạo môi trường ảo `.venv` bằng Python 3.10

## Ô 4. Cài PyTorch bản CUDA

Lệnh dưới đây là lệnh đã dùng thành công trên máy tác giả (torch 2.14.1+cu132). Nếu bản này không còn trên kho hoặc driver của bạn không hỗ trợ CUDA 13.2, hãy lấy lệnh phù hợp ở trang Get Started của pytorch.org.

**Máy không có GPU NVIDIA:** bỏ qua ô này và chạy lệnh `!.venv\Scripts\python.exe -m pip install torch` ở một ô mới (bản CPU). Ô kiểm tra bên dưới sẽ báo GPU là `False`, đây là bình thường.

```python
!.venv\Scripts\python.exe -m pip install torch==2.14.1 --index-url https://download.pytorch.org/whl/cu132
```
